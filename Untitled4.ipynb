```python
### 📌 CLIP Model Setup

import os
from dotenv import load_dotenv
from google.colab import userdata
from transformers import CLIPModel, CLIPProcessor

# ✅ Load API key from Colab Secrets (dotenv not required in Colab)
# load_dotenv()  # Uncomment if running locally with a .env file
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")

# 🎯 Initialize CLIP model and processor for image-text embeddings
clip_model_name = "openai/clip-vit-base-patch32"

# Load the pretrained CLIP model
clip_embedding_model = CLIPModel.from_pretrained(clip_model_name)

# Load the processor (handles tokenization & image preprocessing)
clip_preprocessor = CLIPProcessor.from_pretrained(clip_model_name)

# Set the model to evaluation mode (inference-ready)
clip_embedding_model.eval()

print("✅ CLIP model initialized and ready for embeddings!")
```

```python
### 📌 Embedding Utility Functions

from PIL import Image
import torch
import numpy as np

def generate_image_embedding(image_source):
    """
    🖼️ Generate a CLIP embedding for an image.

    Args:
        image_source (str | PIL.Image.Image): File path or a PIL Image object.

    Returns:
        np.ndarray: Normalized image embedding vector.
    """
    # ✅ Handle both image file paths and PIL Image objects
    if isinstance(image_source, str):
        image = Image.open(image_source).convert("RGB")
    else:
        image = image_source

    # Preprocess image for CLIP
    processed_inputs = clip_preprocessor(images=image, return_tensors="pt")

    # Generate embedding (no gradient tracking needed)
    with torch.no_grad():
        image_embedding = clip_embedding_model.get_image_features(**processed_inputs)
        # Normalize embedding to unit vector
        image_embedding = image_embedding / image_embedding.norm(dim=-1, keepdim=True)

    return image_embedding.squeeze().numpy()


def generate_text_embedding(text_input):
    """
    ✍️ Generate a CLIP embedding for a text string.

    Args:
        text_input (str): The text to embed.

    Returns:
        np.ndarray: Normalized text embedding vector.
    """
    # Tokenize and preprocess text
    processed_inputs = clip_preprocessor(
        text=text_input,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=77  # CLIP's max token length
    )

    # Generate embedding (inference mode)
    with torch.no_grad():
        text_embedding = clip_embedding_model.get_text_features(**processed_inputs)
        # Normalize embedding to unit vector
        text_embedding = text_embedding / text_embedding.norm(dim=-1, keepdim=True)

    return text_embedding.squeeze().numpy()
```

```python
### 📄 PDF Processing Setup

import fitz  # PyMuPDF
from langchain.text_splitter import RecursiveCharacterTextSplitter

# ✅ Path to your PDF file
pdf_file_path = r"/content/AI Training Document.pdf" # Using a raw string to handle backslashes

# 📥 Open the PDF
pdf_document = fitz.open(pdf_file_path)

# 📦 Containers to store extracted data
document_chunks = []        # Stores text chunks from the PDF
document_embeddings = []    # Stores their corresponding embeddings
image_store = {}            # Stores extracted images for further processing or LLM use

# ✂️ Initialize a text splitter to break content into manageable chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,      # Each chunk will have ~500 characters
    chunk_overlap=100    # Overlap between chunks for better context
)

print("✅ PDF loaded and text splitter initialized.")
```

```python
# Display the first few extracted document chunks
if document_chunks:
    print("First 5 document chunks:")
    for i, doc in enumerate(document_chunks[:5]):
        print(f"--- Chunk {i+1} (Page {doc.metadata.get('page', '?')}, Type: {doc.metadata.get('type', 'unknown')}) ---")
        print(doc.page_content)
else:
    print("No document chunks were extracted.")
```

```python
### 📚 Process PDF: Extract Text & Images + Generate CLIP Embeddings

import io
import base64
from PIL import Image
from langchain_core.documents import Document

# ✅ Iterate through each page of the PDF
for page_index, page in enumerate(pdf_document):

    # =============================
    # ✍️ 1. Extract & Embed Text
    # =============================
    raw_text = page.get_text()

    if raw_text.strip():
        # Create a temporary Document for chunking
        temp_doc = Document(
            page_content=raw_text,
            metadata={"page": page_index, "type": "text"}
        )

        # Split long text into smaller chunks for better embedding
        text_chunks = text_splitter.split_documents([temp_doc])

        # Generate CLIP embeddings for each text chunk
        for chunk in text_chunks:
            text_vector = generate_text_embedding(chunk.page_content)
            document_embeddings.append(text_vector)
            document_chunks.append(chunk)

    # =============================
    # 🖼️ 2. Extract & Embed Images
    # =============================
    for image_index, image_info in enumerate(page.get_images(full=True)):
        try:
            xref = image_info[0]
            image_data = pdf_document.extract_image(xref)
            image_bytes = image_data["image"]

            # 🪄 Convert raw bytes to a PIL image
            pil_image = Image.open(io.BytesIO(image_bytes)).convert("RGB")

            # 🔖 Create a unique ID for this image
            image_id = f"page_{page_index}_img_{image_index}"

            # 📦 Store image as base64 (useful for GPT-4V or multimodal LLMs)
            image_buffer = io.BytesIO()
            pil_image.save(image_buffer, format="PNG")
            image_base64 = base64.b64encode(image_buffer.getvalue()).decode()
            image_store[image_id] = image_base64

            # 🤖 Generate CLIP embedding for the image
            image_vector = generate_image_embedding(pil_image)
            document_embeddings.append(image_vector)

            # 🧾 Create a Document object to represent this image
            image_doc = Document(
                page_content=f"[Image: {image_id}]",
                metadata={"page": page_index, "type": "image", "image_id": image_id}
            )
            document_chunks.append(image_doc)

        except Exception as error:
            print(f"⚠️ Error processing image {image_index} on page {page_index}: {error}")
            continue

# ✅ Close the PDF file after processing
pdf_document.close()

print("✅ PDF processing complete! Text and image embeddings are ready.")
```

```python
### 🧠 Create a Unified FAISS Vector Store with CLIP Embeddings

import faiss
import numpy as np

# ✅ Convert collected embeddings into a NumPy array
embedding_matrix = np.array(document_embeddings).astype("float32")
print(f"✅ Total embeddings: {embedding_matrix.shape[0]}, Dimension: {embedding_matrix.shape[1]}")

# 🏗️ Initialize a FAISS index (L2 similarity)
index = faiss.IndexFlatL2(embedding_matrix.shape[1])

# 📥 Add embeddings to the FAISS index
index.add(embedding_matrix)

print("✅ FAISS vector store created and populated with CLIP embeddings!")
```

```python
### 🧠 Create a Custom FAISS Vector Store (Using Precomputed CLIP Embeddings)

from langchain_community.vectorstores import FAISS

# ✅ Pair each document's content with its precomputed embedding
text_embedding_pairs = [
    (doc.page_content, embedding)
    for doc, embedding in zip(document_chunks, embedding_matrix)
]

# 🏗️ Build the FAISS vector store directly from embeddings
clip_vector_store = FAISS.from_embeddings(
    text_embeddings=text_embedding_pairs,
    embedding=None,   # Not needed since we already computed embeddings
    metadatas=[doc.metadata for doc in document_chunks]
)

print("✅ FAISS vector store created successfully!")
clip_vector_store
```

```python
### 🔍 Multimodal Retrieval Function (Text & Image)

def retrieve_similar_chunks(query_text, top_k=5):
    """
    Retrieve the most relevant text and image chunks from the FAISS vector store
    based on a text query using CLIP embeddings.

    Args:
        query_text (str): The input text query.
        top_k (int): Number of top similar chunks to return.

    Returns:
        List[Document]: Top-k most relevant documents (text or image).
    """
    # ✅ Generate CLIP embedding for the query
    query_vector = generate_text_embedding(query_text)

    # 🔎 Perform similarity search in the unified FAISS vector store
    top_results = clip_vector_store.similarity_search_by_vector(
        embedding=query_vector,
        k=top_k
    )

    return top_results
```

```python
### 🔍 Multimodal Retrieval Function (Text & Image)

def retrieve_similar_chunks(query_text, top_k=5):
    """
    Retrieve the most relevant text and image chunks from the FAISS vector store
    based on a text query using CLIP embeddings.

    Args:
        query_text (str): The input text query.
        top_k (int): Number of top similar chunks to return.

    Returns:
        List[Document]: Top-k most relevant documents (text or image).
    """
    # ✅ Generate CLIP embedding for the query
    query_vector = generate_text_embedding(query_text)

    # 🔎 Perform similarity search in the unified FAISS vector store
    top_results = clip_vector_store.similarity_search_by_vector(
        embedding=query_vector,
        k=top_k
    )

    return top_results
```

```python
### 🖼️ Create a Multimodal Message

from langchain.schema import HumanMessage

def create_multimodal_message(query_text, retrieved_documents):
    """
    Create a structured message combining text and images .

    Args:
        query_text (str): The user's query.
        retrieved_documents (List[Document]): Retrieved text and image documents.

    Returns:
        HumanMessage: A multimodal message ready to send .
    """
    message_content = []

    # -------------------------
    # 1️⃣ Add the user's query
    # -------------------------
    message_content.append({
        "type": "text",
        "text": f"Question: {query_text}\n\nContext:\n"
    })

    # -------------------------
    # 2️⃣ Separate text and image documents
    # -------------------------
    text_docs = [doc for doc in retrieved_documents if doc.metadata.get("type") == "text"]
    image_docs = [doc for doc in retrieved_documents if doc.metadata.get("type") == "image"]

    # -------------------------
    # 3️⃣ Add text excerpts
    # -------------------------
    if text_docs:
        text_context = "\n\n".join(
            [f"[Page {doc.metadata['page']}]: {doc.page_content}" for doc in text_docs]
        )
        message_content.append({
            "type": "text",
            "text": f"Text excerpts:\n{text_context}\n"
        })

    # -------------------------
    # 4️⃣ Add images as base64
    # -------------------------
    for doc in image_docs:
        image_id = doc.metadata.get("image_id")
        if image_id and image_id in image_data_store:
            # Add a reference text before the image
            message_content.append({
                "type": "text",
                "text": f"\n[Image from page {doc.metadata['page']}]:\n"
            })
            # Add the image itself in base64
            message_content.append({
                "type": "image_url",
                "image_url": {
                    "url": f"data:image/png;base64,{image_data_store[image_id]}"
                }
            })

    # -------------------------
    # 5️⃣ Add final instruction
    # -------------------------
    message_content.append({
        "type": "text",
        "text": "\n\nPlease answer the question based on the provided text and images."
    })

    return HumanMessage(content=message_content)
```

```python
### 🚀 Multimodal PDF RAG Pipeline

def multimodal_pdf_rag_pipeline(user_query):
    """
    Main pipeline to perform multimodal Retrieval-Augmented Generation (RAG)
    over a PDF using CLIP embeddings.

    Args:
        user_query (str): The question or prompt from the user.

    Returns:
        str: Response based on retrieved text and image context.
    """
    # -------------------------
    # 1️⃣ Retrieve relevant text & image chunks
    # -------------------------
    retrieved_docs = retrieve_similar_chunks(user_query, top_k=5)

    # -------------------------
    # 2️⃣ Format a multimodal message for GPT-4V
    # -------------------------
    multimodal_message = create_multimodal_message(user_query, retrieved_docs)

    # -------------------------
    # 3️⃣ Get the LLM response
    # -------------------------
    response = vision_llm.invoke([multimodal_message])

    # -------------------------
    # 4️⃣ Print retrieved context for reference/debugging
    # -------------------------
    print(f"\nRetrieved {len(retrieved_docs)} documents:")
    for doc in retrieved_docs:
        doc_type = doc.metadata.get("type", "unknown")
        page = doc.metadata.get("page", "?")
        if doc_type == "text":
            preview = doc.page_content[:100] + "..." if len(doc.page_content) > 100 else doc.page_content
            print(f"  - Text from page {page}: {preview}")
        else:
            print(f"  - Image from page {page}")
    print("\n")

    return response.content
```
